```python
import pandas as pd
df = pd.read_csv(r"C:\Users\navin\Downloads\fetal_health.csv")
print(df)
```

```python
# Dataset shape
df.shape
```

```python
# Data types
df.info()
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Check duplicates
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
#It converts the fetal_health column to integer type.
df['fetal_health'] = df['fetal_health'].astype(int)
```

```python
df
```

```python
df['fetal_health']=df['fetal_health'].map({1:0,2:1,3:2})
```

```python
df['fetal_health'].value_counts()
```

```python
# X → all rows, all columns except last
X = df.iloc[:, :-1]
```

```python
# y → all rows, last column
y = df.iloc[:, -1]
```

```python
# Check the number of features
n_features = X.shape[1]
```

```python
X.shape[1]
```

```python
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python

# 4 input layer
# 8 hidden layer
# 3 output layer
# tensorflow model 
#     sequential model 
#     functional model 
#     model subclassing

# Artifical neural networks
model = Sequential()
model.add(Dense(1024, activation='relu', input_shape=(n_features,)))
model.add(Dense(512, activation='relu'))
#model.add(Dense(256, activation='relu'))
#model.add(Dense(16, activation='relu'))
#model.add(Dense(128, activation='relu'))

model.add(Dense(3, activation='softmax'))  # (Iris-setosa, Iris-versicolor, Iris-virginica)
 
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=150, batch_size=32, verbose=2)
```

```python
loss, acc = model.evaluate(X_test, y_test, verbose=0)
print('Test Accuracy: %.3f' % acc)
```

```python
import numpy as np

sample = np.array([[]])  
prediction = model.predict(sample)

predicted_class = np.argmax(prediction, axis=1)
print('Predicted class:', predicted_class)
```
